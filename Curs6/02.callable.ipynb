```python
class Motor:
    def __init__(self, serie):
        self.__serie = serie

    @property
    def serie(self):
        return self.__serie
    
    def __call__(self):
        print("Obiectul a fost chemat...")
    

m1 = Motor("3213")
m1()
```

```python
class Motor:
    def __init__(self, serie):
        self.__serie = serie

    @property
    def serie(self):
        return self.__serie
    
    def __call__(self, noi_km):
        print(f"Obiectul a fost parcurs alti km {noi_km}...")
    

m1 = Motor("3213")
m1(100)
```
